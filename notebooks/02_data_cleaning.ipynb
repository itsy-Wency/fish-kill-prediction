```python
import pandas as pd
import numpy as np

# Load dataset
data = pd.read_csv("../data/processed/combined_pond_dataset.csv")

# Preview
data.head()
```

```python
data.info()
data.describe()
```

```python
data['created_at'] = pd.to_datetime(data['created_at'])

# Sort by time (VERY IMPORTANT for time-series)
data = data.sort_values('created_at')
```

```python
data.isnull().sum()
```

```python
data = data.dropna()
data = data[data['turbidity'] >= 0]
data = data[(data['ph'] >= 0) & (data['ph'] <= 14)]
data = data[data['dissolved_oxygen'] <= 20]
data = data[data['ammonia'] < 10]
data['nitrate'] = np.log1p(data['nitrate'])
data = data[(data['temperature'] >= 0) & (data['temperature'] <= 40)]
```

```python
data = data.reset_index(drop=True)
```

```python
data.describe()
```

```python
data.to_csv("../data/processed/cleaned_dataset.csv", index=False)
```

This data cleaning process ensured that the dataset is accurate, consistent, and suitable for subsequent modeling. Invalid and unrealistic sensor readings—such as negative turbidity values, extreme ammonia levels, and out-of-range temperature measurements—were identified and removed based on domain-specific thresholds. The timestamp variable was properly formatted and ordered to preserve the time-series nature of the data, while skewed variables like nitrate were transformed to stabilize their distribution. As a result, the cleaned dataset now reflects realistic aquaponics conditions and provides a reliable foundation for developing machine learning models for fish kill prediction.
